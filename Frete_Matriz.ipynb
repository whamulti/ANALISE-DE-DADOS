```python
import pandas as pd
df = pd.read_excel("Custo_Frete - MATRIZ (1).xlsx")
df.head()
```

```python
# CALCULOS METRICOS

# CALCULO DA MÉDIA

media = df['FRETE_CUSTO'].mean()
print(f'Media dos Fretes: {media:.2f}')

# CALCULO DA MEDIANA

mediana = df['FRETE_CUSTO'].median()
print(f'Mediana dos Fretes: {mediana:.2f}')

# CALCULO DA MODA

moda = df['FRETE_CUSTO'].mode()
print(f'Moda dos Fretes: {moda}')
print(f'Moda: {[round(m,2)for m in moda.tolist()]}')
```
